```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
df = pd.read_csv("patient_details.csv")
df
```

```python
print(df.isnull().sum())
```

```python
df['name'].fillna("Unknown", inplace=True)
df
```

```python
df['age'].fillna(df['age'].mean(), inplace=True)


df
```

```python
df['height_cm'] = pd.to_numeric(df['height_cm'], errors='coerce')# errors='coerce' means If Python finds a value that cannot be converted into a number, it will replace it with NaN (missing value).
df
```

```python
df['height_cm'].fillna(df['height_cm'].mean(), inplace=True)
df['height_cm'] = df['height_cm'].astype(int)
df
```

```python
df['gender'] = df['gender'].str.upper()

df['gender'] = df['gender'].replace({
    "MALE":"M",
    "FEMALE":"F"
})
df
```

```python
df['city'] = df['city'].str.title()
df
```

```python
df['admission_date'] = pd.to_datetime(df['admission_date'], format='mixed', dayfirst=True)
df
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df.reset_index(drop=True, inplace=True)
df
```

```python
# normalization
mm = MinMaxScaler()
scaled_data = pd.DataFrame(mm.fit_transform(df[['age','height_cm']]))
print(scaled_data)
```

```python
# standardization
ss = StandardScaler()
scaled_data = pd.DataFrame(ss.fit_transform(df[['age','height_cm']]))
print(scaled_data)
```

```python
scaled_data.mean(),scaled_data.std()
```
