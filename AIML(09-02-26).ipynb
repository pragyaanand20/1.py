```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rollno=np.arange(1,51)
height=[]
weight=[]
stipen=[]
name=["Riya","Rahul","Jeevika","Akanksha","Kamal"]*10
for i in range(50):
    a=np.random.randint(150,200)
    b=np.random.randint(30,110)
    c=np.random.randint(500,1000)
    height.append(a)
    weight.append(b)
    stipen.append(c)
n=np.array((rollno,name,height,weight,stipen))
a = np.transpose(n)
df=pd.DataFrame(a,columns=("rollno","name","height","weight","stipen"))
df
```

```python
l=[]
y=df['height'].astype(float)
min=y.min()
max=y.max()
for x in y:
    n=(x-min)/(max-min)
    l.append(n)
print(l)
```

```python
w=[]
y=df['weight'].astype(float)
min=y.min()
max=y.max()
for x in y:
    n=(x-min)/(max-min)
    w.append(n)
print(w)
```
