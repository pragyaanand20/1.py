```python
def dfs(st,goal,graph,vis=None):
    if vis is None:
        vis=set()
    print(st)
    if st==goal:
        return True
    vis.add(st)
    for child in graph[st]:
        if child not in vis:
            if dfs(child,goal,graph,vis):
                return True
            return False
```

```python
graph={'A':['B','C','D'],
       'B':['A','C','E'],
       'C':['A','B','D','E'],
       'D':['A','C','E'],
       'E':['B','C','D']}
dfs('A','D',graph)
```

```python
graph={'A':['B','C'],
       'B':['A','D','F'],
       'C':['A','F'],
       'D':['B'],
       'F':['B','C']}
dfs('A','D',graph)
```

```python
graph={'A':['B','C'],
       'B':['A','C'],
       'C':['A','B']}
dfs('A','C',graph)
```

```python
def bfs(st,goal,graph):
    vis=set()
    queue=[st]
    path=[]
    vis.add(st)
    while queue:
        node=queue.pop(0)
        path.append(node)
        if node==goal:
            return path
        for neighbour in graph[node]:
            if neighbour not in vis:
                vis.add(neighbour)
                queue.append(neighbour)
    return path
```

```python
graph={'A':['B','C','D'],
       'B':['A','C','E'],
       'C':['A','B','D','E'],
       'D':['A','C','E'],
       'E':['B','C','D']}
bfs('A','D',graph)
```

```python
graph={'A':['B','C'],
       'B':['D'],
       'C':['E'],
       'D':[],
       'E':[]}
bfs('A','E',graph)
```
